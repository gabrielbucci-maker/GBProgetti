passiamo un dizioniario con 2 piatti per ogni portata e 2 bevande, il dizionatio **menu** sarà passato come variabile delle funzioni. 

```python
menu = {
    "antipasti": {
        "Bruschetta": {"prezzo": "5€", "intolleranze": ["glutine"], "ingredienti": ["pane", "pomodoro", "aglio"]},
        "Caprese": {"prezzo": "7€", "intolleranze": [], "ingredienti": ["mozzarella", "pomodoro", "basilico"]}
    },
    "primi": {
        "Carbonara": {"prezzo": "12€", "intolleranze": ["uova", "latticini"], "ingredienti": ["pasta", "uova", "guanciale"]},
        "Risotto ai funghi": {"prezzo": "14€", "intolleranze": ["latticini"], "ingredienti": ["riso", "funghi", "parmigiano"]}
    },
    "secondi": {
        "Bistecca": {"prezzo": "18€", "intolleranze": [], "ingredienti": ["carne", "sale", "olio"]},
        "Pollo al forno": {"prezzo": "15€", "intolleranze": [], "ingredienti": ["pollo", "patate", "rosmarino"]}
    },
    "drink": {
        "Mojito": {"prezzo": "8€", "intolleranze": ["alcol"], "ingredienti": ["rum", "zucchero", "menta", "lime"]},
        "Spritz": {"prezzo": "7€", "intolleranze": ["alcol"], "ingredienti": ["aperol", "prosecco", "soda"]}
    }
}
```

**aggiungi_piatto** funzione che ci permette di selezionare a quale categoria aggiungerelo, mettere tutti gli attributi richiesti e salvarli nel dizionario. **Stampando** il menu aggiornato

Contiene una **doppia verifica** sulla sezione se è esistente nel menu, e se il piatto è gia esistente in quella sezione.

```python
# Funzione per aggiungere un piatto al menu
def aggiungi_piatto(menu):
    # Chiediamo la sezione del menu
    sezione = input("In quale sezione vuoi aggiungere un piatto? (antipasti, primi, secondi, drink): ").strip().lower()

    # Controllo che la sezione esista
    if sezione not in menu:
        print("Sezione non valida! Riprova.")
        return
    # Raccolta delle informazioni del piatto
    nome = input("Nome del piatto: ").strip()
    
    # verifica se il piatto è gia presente
    if nome in menu[sezione] :
        print(" ")
        print(f"Il piatto '{nome}' è già presente nella categoria '{sezione}'.")
        return
    prezzo = input("Prezzo del piatto (€): ").strip()
    intolleranze = input("Intolleranze alimentari (separate da virgola, lascia vuoto se nessuna): ").strip()
    ingredienti = input("Ingredienti (separati da virgola): ").strip()
     # Creazione del dizionario piatto
    piatto = {
            "prezzo": prezzo,
            "intolleranze": intolleranze.split(",") if intolleranze else [],
            "ingredienti": ingredienti.split(",")
    }
    # Aggiunta del piatto alla categoria corrispondente
    menu[sezione][nome] = piatto

    print(f"\n✅ Piatto '{nome}' aggiunto con successo alla sezione {sezione}!")


aggiungi_piatto(menu)  # Per aggiungere un nuovo piatto
stampa_menu(menu)  # Per vedere il menu aggiornato
```

**Stampa_menu** funzione che ci peremette di stampare tutto il menu attraverso 2 cicli

```python
# Funzione per stampare il menu aggiornato
def stampa_menu(menu):
    for sezione, piatti in menu.items():
        print(f"\n--- {sezione.capitalize()} ---")
        for nome, dettagli in piatti.items():
            print(f"🍽 {nome}: {dettagli['prezzo']}€")
            print(f"   Intolleranze: {', '.join(dettagli['intolleranze']) if dettagli['intolleranze'] else 'Nessuna'}")
            print(f"   Ingredienti: {', '.join(dettagli['ingredienti'])}")
        print("-" * 40)
```

**ricerca_ingredienti** ci permette di cercare uno specifico ingrediente in tutto il dizionario

```python
def ricerca_ingredienti(menu):
    ingrediente = input("Quale ingrediente vuoi cercare: ").strip().lower()
    trovati = []

    for sezione in menu:
        for piatto, info in menu[sezione].items():
            ingredienti_piatto = [i.lower() for i in info["ingredienti"]]
            if ingrediente in ingredienti_piatto:
                trovati.append((piatto, sezione))

    if not trovati:
        print("❌ L'ingrediente non è presente in nessun piatto del menu.")
        return

    print(f"\n--- Piatti con l'ingrediente {ingrediente.capitalize()} ---")
    for piatto, sezione in trovati:
        print(f"🍽 {piatto} (sezione: {sezione})")
ricerca_ingredienti(menu)
```

**elimina_piatto** ci permette di eliminiare un piatto dal dizionario menu attraverso la scelta della sezione -> il controllo se la fuznione esiste -> la stampa dei piatti della sezione per mostrare quali PUOI ELIMINARE, ti fa inserire quale vuoi eliminare -> ultimo controllo se il piatto esiste -> elimino.

```python
def elimina_piatto(menu):
    # Chiediamo la sezione del menu
    sezione = input("Da quale sezione vuoi eliminare un piatto? (antipasti, primi, secondi, drink): ").strip().lower()

    # Controllo se la sezione esiste
    if sezione not in menu:
        print("❌ Sezione non valida! Riprova.")
        return

    # Stampiamo i piatti disponibili in quella sezione
    print(f"\n--- Piatti disponibili in {sezione.capitalize()} ---")
    for piatto in menu[sezione]:
        print(f"🍽 {piatto}")

    # Chiediamo quale piatto eliminare
    nome_piatto = input("\nInserisci il nome del piatto da eliminare: ").strip()

    # Controlliamo se il piatto esiste e lo eliminiamo
    if nome_piatto in menu[sezione]:
        del menu[sezione][nome_piatto]
        print(f"✅ Piatto '{nome_piatto}' eliminato con successo dalla sezione {sezione}!")
    else:
        print(f"❌ Il piatto '{nome_piatto}' non esiste nella sezione {sezione}. Riprova.")

# Test della funzione
elimina_piatto(menu)  # Per eliminare un piatto
stampa_menu(menu)  # Per vedere il menu aggiornato
```

**filtraggio** permette a scelta dell' utente di fare una scrematura dei piatti che sono all' interno del dizionario per sezione o per prezzo o per entrambi

```python
def filtraggio(menu):
    filtra = input("Secondo quale criterio vuoi filtrare? (sezione, prezzo, entrambi): ").strip().lower()
    
    if filtra not in ["sezione", "prezzo", "entrambi"]:
        print("❌ Opzione non valida! Riprova.")
        return

    if filtra == "sezione" or filtra == "entrambi":
        sezione = input("Quale sezione vuoi vedere? (antipasti, primi, secondi, drink): ").strip().lower()
        if sezione not in menu:
            print("❌ Sezione non valida! Riprova.")
            return

    if filtra == "prezzo" or filtra == "entrambi":
        try:
            prezzom = float(input("Vuoi che il prezzo sia inferiore a (€): ").strip())
        except ValueError:
            print("❌ Inserisci un numero valido!")
            return

    print("\n--- Risultati ---")
    
    for sez, piatti in menu.items():  
        # Se il filtro è solo per sezione, mostra solo quella sezione
        if filtra == "sezione" and sez != sezione:
            continue  

        print(f"\n--- {sez.capitalize()} ---")
        for nome, dettagli in piatti.items():
            # Rimuoviamo il simbolo dell'euro e convertiamo in float il prezzo
            prezzo_num = float(dettagli['prezzo'].replace("€", "").strip())  
            
            # Se il filtro è per entrambi, controlliamo sia la sezione che il prezzo
            if filtra == "entrambi":
                if sez != sezione or prezzo_num >= prezzom:
                    continue  # Salta se non soddisfa entrambi i criteri

            # Se il filtro è solo per prezzo, controlliamo solo il prezzo
            if filtra == "prezzo" and prezzo_num >= prezzom:
                continue  # Salta se il prezzo è maggiore o uguale al valore inserito

            # Stampa i piatti che passano il filtro
            print(f"🍽 {nome}: {dettagli['prezzo']}€")
            print(f"   Intolleranze: {', '.join(dettagli['intolleranze']) if dettagli['intolleranze'] else 'Nessuna'}")
            print(f"   Ingredienti: {', '.join(dettagli['ingredienti'])}")
            print("-" * 40)
```

Questa funzione ci mostra quale è l' ingrediente più usato all' interno del menu, con **3 cicli** che ci permettono di arrivare agli ingrediente e leggerli di goni piatto e di ogni sezione, e attraferso un **if**  con il suo interno un conteggio per ogni volta che l' ingrediente viene trovato

```python
def ingrediente_piu_usato(menu):
    conteggio = {}

    for sezione in menu:
        for piatto in menu[sezione].values():
            for ingrediente in piatto["ingredienti"]:
                ingrediente = ingrediente.lower()
                if ingrediente in conteggio:
                    conteggio[ingrediente] += 1
                else:
                    conteggio[ingrediente] = 1

    # Trova l'ingrediente con il valore massimo
    ingrediente_max = None
    max_occorrenze = 0

    for ingrediente, occorrenze in conteggio.items():
        if occorrenze > max_occorrenze:
            max_occorrenze = occorrenze
            ingrediente_max = ingrediente

    print(f"🥇 L'ingrediente più usato è '{ingrediente_max}' con {max_occorrenze} occorrenze.")
ingrediente_piu_usato(menu)
```

La funzione sotto stante ci permette di trovare il **piatto** con **il maggior numero** di intolleranze. 

```python
def piatto_con_piu_intolleranze(menu):
    max_intolleranze = -1
    piatto_risultato = ""
    sezione_risultato = ""

    for sezione in menu:
        for nome_piatto, info in menu[sezione].items():
            num = len(info["intolleranze"])
            if num > max_intolleranze:
                max_intolleranze = num
                piatto_risultato = nome_piatto
                sezione_risultato = sezione

    if max_intolleranze == 0:
        print("✅ Nessun piatto ha intolleranze!")
    else:
        print(f"⚠️ Il piatto con più intolleranze è '{piatto_risultato}' (sezione: {sezione_risultato}) con {max_intolleranze} intolleranze.")
piatto_con_piu_intolleranze(menu)
```

la funzione **prezzo_medio** calcola il prezzo medio del medio

```python
def prezzo_medio(menu):
    totale = 0
    count = 0

    for sezione in menu:
        for piatto in menu[sezione].values():
            prezzo_str = piatto["prezzo"].replace("€", "").strip()
            try:
                totale += float(prezzo_str)
                count += 1
            except ValueError:
                print(f"Errore nel convertire il prezzo: {prezzo_str}")

    if count == 0:
        print("❌ Nessun piatto trovato.")
    else:
        media = totale / count
        print(f"💰 Il prezzo medio dei piatti è: {media:.2f}€")
prezzo_medio(menu)
```

funzione che raggruppa le tre funzione di statistiche **prezzo medio**, **piatto con piu intolleranze**, **ingrediente più usato**.

```python
def statistiche (menu):
    ingrediente_piu_usato(menu)
    print("-" * 40)
    piatto_con_piu_intolleranze(menu)
    print("-" * 40)
    prezzo_medio(menu)
statistiche(menu)
```
